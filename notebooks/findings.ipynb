# Population capture in endometriosis and PCOS interventional drug trials

The extraction `population` block is a publishable finding on its own: it measures how often registry records state the female-specific variables that would let a trial actually falsify a mechanism.

Numbers below are computed from the exported snapshot (`dashboard/public/data/snapshot.json`), which was built from a live ClinicalTrials.gov v2 ingest.

```python
import json
from pathlib import Path

snap = json.loads(Path("../dashboard/public/data/snapshot.json").read_text())
rates = snap["landscape"]["population_capture_rates"]
print(f"n_assets = {snap['landscape']['n_assets']}")
print(f"mean capture = {snap['landscape']['mean_female_specific_capture']:.1%}\n")
for field, row in rates.items():
    print(f"{field:42} {row['rate']:6.1%}  {row['captured']}/{row['n']}")
```

## How to read this

- **Age range** is almost always present because ClinicalTrials.gov has structured min/max age fields.
- **Diagnosis method** is often in free-text eligibility (laparoscopy, Rotterdam).
- **Disease severity / ASRM stage** and **prior treatment history** are the usual holes. A pain trial that does not say whether it enrolled rASRM I–II or IV, or whether patients had failed a GnRH analogue, cannot have told you the biology failed.
- That is why stated efficacy plus two or fewer of these variables classifies as `efficacy_uninterpretable` — the highest-value category in the triage, not a compliment to the protocol.
